# Phân loại ảnh CIFAR-100 bằng PyTorch

Notebook này huấn luyện một mạng CNN để phân loại ảnh vào 100 lớp CIFAR-100. Dữ liệu được tải tự động qua `torchvision`. Nếu máy có GPU NVIDIA và cài bản PyTorch hỗ trợ CUDA, notebook sẽ tự chọn GPU.

```python
# N?u thi?u th? vi?n, b? d?u # v? ch?y cell n?y, sau ?? kh?i ??ng l?i kernel.
# %pip install torch torchvision matplotlib scikit-learn

import random
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('PyTorch:', torch.__version__)
print('Device:', device)
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
import matplotlib.pyplot as plt
import matplotlib as mpl

# Cấu hình Matplotlib sử dụng font hỗ trợ tiếng Việt trên Windows/macOS/Linux
plt.rcParams['font.sans-serif'] = ['Segoe UI', 'Arial', 'Times New Roman', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False  # Tránh lỗi hiển thị dấu trừ (-)
```

## Tải và chia dữ liệu CIFAR-100

Tập huấn luyện ban đầu có 50.000 ảnh, được chia thành 45.000 ảnh train và 5.000 ảnh validation. Tập test có 10.000 ảnh. Chỉ tập train dùng crop và lật ngang ngẫu nhiên; validation và test chỉ được chuẩn hóa.

```python
DATA_DIR = './data'
BATCH_SIZE = 128

mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)
train_transform = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(mean, std),
])
eval_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean, std),
])

# T?i b? d? li?u l?n ??u c?n k?t n?i Internet.
train_augmented = datasets.CIFAR100(DATA_DIR, train=True, download=True, transform=train_transform)
train_unaugmented = datasets.CIFAR100(DATA_DIR, train=True, download=False, transform=eval_transform)
test_dataset = datasets.CIFAR100(DATA_DIR, train=False, download=True, transform=eval_transform)
class_names = train_augmented.classes

generator = torch.Generator().manual_seed(SEED)
indices = torch.randperm(len(train_augmented), generator=generator).tolist()
train_indices, val_indices = indices[5000:], indices[:5000]
train_dataset = Subset(train_augmented, train_indices)
val_dataset = Subset(train_unaugmented, val_indices)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0, pin_memory=torch.cuda.is_available())
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=torch.cuda.is_available())
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=torch.cuda.is_available())
print(f'Train: {len(train_dataset)}, validation: {len(val_dataset)}, test: {len(test_dataset)}')
print('S? l?p:', len(class_names))
```

```python
# Hiển thị ảnh mẫu sau khi hoàn nguyên bước chuẩn hóa.
def show_tensor_image(image):
    image = image.cpu().permute(1, 2, 0).numpy()
    image = np.clip(image * np.array(std) + np.array(mean), 0, 1)
    return image

images, labels = next(iter(train_loader))
fig, axes = plt.subplots(3, 3, figsize=(8, 8))
for i, ax in enumerate(axes.flat):
    ax.imshow(show_tensor_image(images[i]))
    ax.set_title(class_names[labels[i].item()])
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## CNN tự xây dựng

Mô hình gồm ba khối tích chập với lần lượt 96, 192 và 384 kênh. Mỗi khối có hai lớp convolution, BatchNorm, ReLU, MaxPool và Dropout; phần phân loại xuất logits cho 100 lớp.


```python
class ConvBlock(nn.Module):
    def __init__(self, in_channels, out_channels, dropout):
        super().__init__()
        self.block = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),
            nn.Dropout2d(dropout),
        )

    def forward(self, x):
        return self.block(x)


class CustomCNN(nn.Module):
    def __init__(self, num_classes=100):
        super().__init__()
        self.features = nn.Sequential(
            ConvBlock(3, 96, dropout=0.10),
            ConvBlock(96, 192, dropout=0.15),
            ConvBlock(192, 384, dropout=0.20),
        )
        self.classifier = nn.Sequential(
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),
            nn.Linear(384, 384),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(384, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)


model = CustomCNN(num_classes=100).to(device)
print(model)
print('Trainable parameters:', sum(p.numel() for p in model.parameters() if p.requires_grad))
```

## Huấn luyện mô hình

`CrossEntropyLoss` nhận logits trực tiếp nên lớp cuối không dùng Softmax. Notebook lưu checkpoint có validation loss thấp nhất và dừng sớm nếu validation loss không cải thiện trong 10 epoch liên tiếp.

```python
EPOCHS = 80
PATIENCE = 10
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.5,
    patience=3,
    min_lr=1e-5,
)

history = {'train_loss': [], 'train_accuracy': [], 'val_loss': [], 'val_accuracy': []}
best_val_loss = float('inf')
epochs_without_improvement = 0
model_path = 'cnn_cifar100_custom_wide.pth'

for epoch in range(EPOCHS):
    model.train()
    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0
    for inputs, targets in train_loader:
        inputs, targets = inputs.to(device), targets.to(device)
        optimizer.zero_grad(set_to_none=True)
        logits = model(inputs)
        loss = criterion(logits, targets)
        loss.backward()
        optimizer.step()
        train_loss_sum += loss.item() * inputs.size(0)
        train_correct += (logits.argmax(dim=1) == targets).sum().item()
        train_total += targets.size(0)

    model.eval()
    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0
    with torch.no_grad():
        for inputs, targets in val_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            logits = model(inputs)
            loss = criterion(logits, targets)
            val_loss_sum += loss.item() * inputs.size(0)
            val_correct += (logits.argmax(dim=1) == targets).sum().item()
            val_total += targets.size(0)

    train_loss = train_loss_sum / train_total
    train_accuracy = train_correct / train_total
    val_loss = val_loss_sum / val_total
    scheduler.step(val_loss)
    val_accuracy = val_correct / val_total
    history['train_loss'].append(train_loss)
    history['train_accuracy'].append(train_accuracy)
    history['val_loss'].append(val_loss)
    history['val_accuracy'].append(val_accuracy)
    print(f'Epoch {epoch + 1:02d}/{EPOCHS} | train loss {train_loss:.4f}, acc {train_accuracy:.3%} | val loss {val_loss:.4f}, acc {val_accuracy:.3%}')

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        epochs_without_improvement = 0
        torch.save(model.state_dict(), model_path)
    else:
        epochs_without_improvement += 1
        if epochs_without_improvement >= PATIENCE:
            print('Early stopping.')
            break

model.load_state_dict(torch.load(model_path, map_location=device))
print('Loaded best checkpoint:', model_path)
```

## PCA - TRUC QUAN HOA DAC TRUNG CNN KERNEL 3x3

```python
# ============================================================
# PCA - TRUC QUAN HOA DAC TRUNG CNN KERNEL 3x3
# Dataset: CIFAR-100
# ============================================================

import os
import numpy as np
import matplotlib.pyplot as plt
import torch

from torch.utils.data import DataLoader
from sklearn.decomposition import PCA

# ------------------------------------------------------------
# 1. Khoi tao mo hinh kernel 3x3 va nap trong so
# ------------------------------------------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = CustomCNN(num_classes=100).to(device)

checkpoint_path = "cnn_cifar100_custom_wide.pth"

if not os.path.exists(checkpoint_path):
    raise FileNotFoundError(
        f"Khong tim thay file trong so: {checkpoint_path}. "
        "Hay thay bang dung duong dan file .pth kernel 3x3."
    )

checkpoint = torch.load(checkpoint_path, map_location=device)

if isinstance(checkpoint, dict) and "state_dict" in checkpoint:
    model.load_state_dict(checkpoint["state_dict"])
elif isinstance(checkpoint, dict) and "model_state_dict" in checkpoint:
    model.load_state_dict(checkpoint["model_state_dict"])
else:
    model.load_state_dict(checkpoint)

model.eval()
print("Da nap trong so mo hinh.")

# ------------------------------------------------------------
# 2. Cau hinh trich xuat dac trung
# ------------------------------------------------------------

MAX_SAMPLES = 3000
BATCH_SIZE = 128
RANDOM_STATE = 42

# Yeu cau test_dataset da duoc tao tu cac cell truoc
test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

features_list = []
labels_list = []

# ------------------------------------------------------------
# 3. Trich xuat dac trung tu CNN
# ------------------------------------------------------------

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)

        # Dac trung sau cac lop convolution
        x = model.features(images)

        # Global Average Pooling va Flatten
        x = model.classifier[0](x)
        x = torch.flatten(x, start_dim=1)

        features_list.append(x.cpu().numpy())
        labels_list.append(labels.numpy())

X = np.concatenate(features_list, axis=0)
y = np.concatenate(labels_list, axis=0)

print("Kich thuoc dac trung:", X.shape)
print("Kich thuoc nhan:", y.shape)

# ------------------------------------------------------------
# 4. Lay toi da MAX_SAMPLES mau de ve
# ------------------------------------------------------------

if len(y) > MAX_SAMPLES:
    rng = np.random.default_rng(RANDOM_STATE)
    indices = rng.choice(
        len(y),
        size=MAX_SAMPLES,
        replace=False
    )
    X = X[indices]
    y = y[indices]

print("So mau dung cho PCA:", len(y))

# ------------------------------------------------------------
# 5. PCA: giam chieu dac trung xuong 2D
# ------------------------------------------------------------

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

explained_variance = pca.explained_variance_ratio_

print(f"PC1: {explained_variance[0] * 100:.2f}%")
print(f"PC2: {explained_variance[1] * 100:.2f}%")
print(
    f"Tong phuong sai duoc giai thich: "
    f"{explained_variance.sum() * 100:.2f}%"
)

# ------------------------------------------------------------
# 6. Ve bieu do PCA
# ------------------------------------------------------------

plt.figure(figsize=(12, 9))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=y,
    cmap="nipy_spectral",
    s=12,
    alpha=0.7
)

cbar = plt.colorbar(scatter)
cbar.set_label("Ma lop CIFAR-100")

plt.xlabel(
    f"Principal Component 1 "
    f"({explained_variance[0] * 100:.2f}%)"
)
plt.ylabel(
    f"Principal Component 2 "
    f"({explained_variance[1] * 100:.2f}%)"
)
plt.title("PCA - CNN Feature Visualization (Kernel 3x3)")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

## t-SNE - CNN Feature Visualization (Kernel 3x3)

```python

import numpy as np
import matplotlib.pyplot as plt

from torch.utils.data import DataLoader
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

# ------------------------------------------------------------
# 1. Cau hinh
# ------------------------------------------------------------
MAX_SAMPLES = 2000
BATCH_SIZE = 128
RANDOM_STATE = 42

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

# ------------------------------------------------------------
# 2. Trich xuat dac trung CNN
# ------------------------------------------------------------
features_list = []
labels_list = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)

        # Qua cac lop convolution
        x = model.features(images)

        # Global Average Pooling: (B, 384, 1, 1)
        x = model.classifier[0](x)

        # Flatten: (B, 384)
        x = torch.flatten(x, start_dim=1)

        features_list.append(x.cpu().numpy())
        labels_list.append(labels.numpy())

X = np.concatenate(features_list, axis=0)
y = np.concatenate(labels_list, axis=0)

print("Kich thuoc dac trung:", X.shape)

# ------------------------------------------------------------
# 3. Chon toi da MAX_SAMPLES mau
# ------------------------------------------------------------
if len(y) > MAX_SAMPLES:
    rng = np.random.default_rng(RANDOM_STATE)
    indices = rng.choice(
        len(y),
        size=MAX_SAMPLES,
        replace=False
    )
    X = X[indices]
    y = y[indices]

print("So mau dung cho t-SNE:", len(y))

# ------------------------------------------------------------
# 4. Chuan hoa dac trung
# ------------------------------------------------------------
X_scaled = StandardScaler().fit_transform(X)

# ------------------------------------------------------------
# 5. Giam chieu bang t-SNE
# ------------------------------------------------------------
tsne = TSNE(
    n_components=2,
    perplexity=30,
    init="pca",
    learning_rate="auto",
    max_iter=1000,
    random_state=RANDOM_STATE
)

X_tsne = tsne.fit_transform(X_scaled)

# ------------------------------------------------------------
# 6. Ve bieu do
# ------------------------------------------------------------
plt.figure(figsize=(12, 9))

scatter = plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=y,
    cmap="nipy_spectral",
    vmin=0,
    vmax=99,
    s=14,
    alpha=0.75
)

cbar = plt.colorbar(scatter)
cbar.set_label("Ma lop CIFAR-100")

plt.xlabel("t-SNE dimension 1")
plt.ylabel("t-SNE dimension 2")
plt.title("t-SNE - CNN Feature Visualization (Kernel 3x3)")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# ============================================================
# VISUALIZE KERNEL 3x3, FEATURE MAPS VA FEATURE VECTOR
# Yêu cau: model, device, test_dataset, class_names, mean, std
# ============================================================

import torch
import numpy as np
import matplotlib.pyplot as plt

model.eval()

# Chọn một ảnh từ tập test
sample_idx = 0
image_tensor, true_label = test_dataset[sample_idx]
input_batch = image_tensor.unsqueeze(0).to(device)

# ------------------------------------------------------------
# 1. TRỰC QUAN HOA KERNEL 3x3 DA HOC
# ------------------------------------------------------------
# Lớp Conv đầu tiên: 96 bộ lọc, mỗi bộ lọc có 3 kênh RGB
conv1 = model.features[0].block[0]
weights = conv1.weight.detach().cpu()

print("Shape kernel:", tuple(weights.shape))
# Kết quả: (96, 3, 3, 3)

# Hiển thị 8 kernel đầu tiên dưới dạng RGB
fig, axes = plt.subplots(2, 4, figsize=(10, 5))

for i, ax in enumerate(axes.flat):
    kernel = weights[i].permute(1, 2, 0).numpy()

    # Chuẩn hóa riêng từng kernel để dễ quan sát
    k_min, k_max = kernel.min(), kernel.max()
    kernel_vis = (kernel - k_min) / (k_max - k_min + 1e-8)

    ax.imshow(kernel_vis)
    ax.set_title(f"Kernel {i + 1}: 3x3")
    ax.axis("off")

plt.suptitle("8 kernel đầu tiên của Conv2d (3 -> 96)")
plt.tight_layout()
plt.show()


# Hiển thị riêng các kênh R, G, B của một kernel
kernel_id = 0
fig, axes = plt.subplots(1, 3, figsize=(8, 3))

for c, ax in enumerate(axes):
    channel = weights[kernel_id, c].numpy()
    vmax = np.abs(channel).max() + 1e-8

    im = ax.imshow(
        channel, cmap="coolwarm",
        vmin=-vmax, vmax=vmax,
        interpolation="nearest"
    )
    ax.set_title(["Red", "Green", "Blue"][c])
    ax.set_xticks(range(3))
    ax.set_yticks(range(3))
    plt.colorbar(im, ax=ax, fraction=0.046)

plt.suptitle(f"Trọng số kernel {kernel_id + 1}: từng kênh RGB")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 2. TRỰC QUAN HOA FEATURE MAP SAU CONVOLUTION
# ------------------------------------------------------------
feature_outputs = {}

def save_feature(name):
    def hook(module, inputs, output):
        feature_outputs[name] = output.detach().cpu().clone()
    return hook

# Hook vào lớp Conv đầu tiên của mỗi block
layers = {
    "ConvBlock 1": model.features[0].block[0],
    "ConvBlock 2": model.features[1].block[0],
    "ConvBlock 3": model.features[2].block[0],
}

handles = []
for name, layer in layers.items():
    handles.append(layer.register_forward_hook(save_feature(name)))

with torch.no_grad():
    logits = model(input_batch)

for handle in handles:
    handle.remove()

# Vẽ 8 feature maps đầu tiên của từng lớp Conv
fig, axes = plt.subplots(3, 8, figsize=(16, 7))

for row, (name, feature) in enumerate(feature_outputs.items()):
    fmap = feature[0]  # (C, H, W)

    for col in range(8):
        ax = axes[row, col]
        ax.imshow(fmap[col].numpy(), cmap="viridis")
        ax.axis("off")

        if row == 0:
            ax.set_title(f"Map {col + 1}", fontsize=9)

    axes[row, 0].set_ylabel(
        f"{name}\n{tuple(fmap.shape)}",
        fontsize=10
    )

plt.suptitle("Feature maps sau Conv2d (trước BatchNorm và ReLU)")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 3. TRỰC QUAN HOA FEATURE VECTOR 384 CHIEU
# ------------------------------------------------------------
# Lấy đầu ra sau AdaptiveAvgPool2d, trước Flatten
pooled_features = {}

def save_pooled(module, inputs, output):
    pooled_features["vector"] = output.detach().cpu().clone()

handle = model.classifier[0].register_forward_hook(save_pooled)

with torch.no_grad():
    logits = model(input_batch)

handle.remove()

# Shape: (1, 384, 1, 1)
feature_vector = pooled_features["vector"].flatten().numpy()

print("Feature vector shape:", feature_vector.shape)
print("Nhãn thật:", class_names[true_label])
print("Lớp dự đoán:", class_names[logits.argmax(dim=1).item()])

fig, axes = plt.subplots(2, 1, figsize=(14, 6))

# Biểu diễn toàn bộ 384 giá trị
axes[0].plot(np.arange(384), feature_vector, linewidth=1)
axes[0].set_title("Feature vector 384 chiều")
axes[0].set_xlabel("Chỉ số đặc trưng")
axes[0].set_ylabel("Giá trị")
axes[0].grid(alpha=0.3)

# Heatmap để nhìn tổng quan các giá trị
im = axes[1].imshow(
    feature_vector.reshape(1, -1),
    aspect="auto", cmap="viridis"
)
axes[1].set_title("Heatmap của feature vector")
axes[1].set_xlabel("Chỉ số đặc trưng")
axes[1].set_yticks([])
plt.colorbar(im, ax=axes[1], orientation="vertical")

plt.tight_layout()
plt.show()
```

## Đánh giá trên tập test

```python
model.eval()
y_true, y_pred = [], []
with torch.no_grad():
    for inputs, targets in test_loader:
        logits = model(inputs.to(device))
        y_pred.extend(logits.argmax(dim=1).cpu().tolist())
        y_true.extend(targets.tolist())

test_accuracy = np.mean(np.array(y_true) == np.array(y_pred))
print(f'Test accuracy: {test_accuracy:.2%}')
print(classification_report(y_true, y_pred, target_names=class_names, zero_division=0))
```
